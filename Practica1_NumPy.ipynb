## Ejercicio 3
Define una función `resumen(lista)` que retorne un diccionario con: longitud, suma, promedio, mínimo y
máximo de la lista.

```python
def resumen(lista):
    '''Retorna longitud, suma, promedio, mínimo y máximo de una lista.'''
    suma = 0
    for x in lista:
        suma = suma + x
    return {
        'longitud': len(lista),
        'suma': suma,
        'promedio': suma / len(lista),
        'minimo': min(lista),
        'maximo': max(lista)
    }

estadisticas = resumen([4, 8, 15, 16, 23, 42])

for clave, valor in estadisticas.items():
    print(f'{clave:12s} → {valor}')
```

## Ejercicio 4
Usa `map()` para aplicar la función `potencia(x, 3)` a la lista `[1, 2, 3, 4, 5]`.

```python
def potencia(base, exponente=2):
    '''Eleva base a exponente. Por defecto calcula el cuadrado.'''
    return base ** exponente

numeros = [1, 2, 3, 4, 5]
cubos = list(map(lambda x: potencia(x, 3), numeros))
print(cubos)
```

## Ejercicio 4
Demuestra la diferencia vista vs. copia: modifica un elemento en un slice y verifica que el array
original también cambia.

```python
original = np.array([1, 2, 3, 4, 5, 6])

vista = original[0:3]
copia = original[0:3].copy()

vista[0] = 999
copia[1] = 888

print('Original:', original)
print('Vista:   ', vista)
print('Copia:   ', copia)
print('¿La vista comparte memoria?', np.shares_memory(original, vista))
print('¿La copia comparte memoria?', np.shares_memory(original, copia))
```

**Resultado:** al modificar la vista, el array original también cambió (su primer elemento pasó a 999),
porque el slicing en NumPy devuelve una vista que comparte memoria con el original. La copia creada con
`.copy()` es independiente, por eso el 888 no aparece en el original.

## Ejercicio 2
Usando broadcasting:

a) Crea un vector fila `[1,2,3,4]` y un vector columna `[[10],[20],[30]]`.

b) Suma ambos vectores y explica el shape del resultado.

```python
fila = np.array([1, 2, 3, 4])
columna = np.array([[10], [20], [30]])

print('Shape fila:   ', fila.shape)
print('Shape columna:', columna.shape)
```

```python
suma = columna + fila
print(suma)
print('Shape del resultado:', suma.shape)
```

**Explicación:** el vector fila tiene shape `(4,)` y el vector columna `(3, 1)`. Según las reglas de
broadcasting, que se aplican de derecha a izquierda:

1. El vector fila tiene menos ejes, así que se le agrega un 1 a la izquierda: `(4,)` → `(1, 4)`.
2. Las dimensiones de tamaño 1 se estiran para coincidir con la otra: `(3, 1)` y `(1, 4)` → `(3, 4)`.

Por eso el resultado tiene shape `(3, 4)`: 3 filas (una por cada valor de la columna) y 4 columnas (una
por cada valor de la fila). Internamente NumPy no copia los datos para lograrlo.
